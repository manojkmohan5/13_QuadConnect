## Step 1.4: Evaluation metrics

Every reply is scored on four metrics. The scorer below is our self-evaluation written as code: the same rules for
all 375 replies, so the scores are reproducible and anyone can check how a number was reached.

| Metric | Question it answers | How it is measured (0 to 1) |
|---|---|---|
| **Accuracy** | Was the answer correct? | Against the reference. Extraction: F1 over the filters the reply sets (each interest, college, connection, day, time and energy is one filter; a missing filter and an invented one both cost, and `{}` is right only when nothing was asked). Classification: the decision. Prediction: 0.6 × rating (exact 1, off by one 0.5) + 0.4 × stay-connected. Summarization: the share of required facts mentioned. Text generation: half the share of questions about a shared interest, half the share of shared interests covered. |
| **Instruction adherence** | Did the model follow the rules it was given? | The share of that type's explicit rules obeyed: allowed keys and values, label set, word limits, number of questions, question marks, no names, no personal-detail questions, no text outside the requested output. |
| **Structure** | Did the output follow the expected format? | 1 for exactly the requested shape (a bare JSON object with the right keys; one plain sentence; exactly three lines numbered 1–3), 0.5 for the right content in the wrong wrapping (JSON inside a code fence or chat, two sentences, unnumbered questions), 0 for no usable output. |
| **Clarity** | Is the output understandable and usable as is? | The share of: no preamble or sign-off ("Sure! Here is..."), not cut off, no repeated lines, no template tokens or garbled characters, and for the two text types, sentences of readable length (30 words or fewer on average). |

**Self-evaluation score (1–10)** = 1 + 9 × accuracy × (0.4 + 0.6 × form), where form is the mean of adherence,
structure and clarity.

Accuracy gates the score, because a wrong answer is useless however neat it looks: a well-formatted `APPROVE` for a
bar scores 1, not 6. Form then decides how much of a correct answer's value it keeps: a correct answer in exactly
the requested form scores 10, and a correct answer in a broken form still keeps 40% (about 4.6), because a person
could use it but QuadConnect's code could not. A perfect reply scores 10.0 and an empty one 1.0. The self-check
below tries the scorer on hand-written replies before any model runs.

````python
CHATTER = re.compile(r"^\s*(sure|certainly|of course|okay|ok\b|here (is|are|'s)|answer\s*:|summary\s*:|"
                     r"response\s*:|output\s*:|json\s*:)|let me know|hope this helps|\bi hope\b", re.IGNORECASE)
JUNK = re.compile(r"<\||\|>|\[/?INST\]|</?s>|<think>|</think>|�|^(assistant|user|system)\s*:",
                  re.IGNORECASE | re.MULTILINE)
SENSITIVE = re.compile(r"\bphone\b|\baddress\b|where (do|does) (you|they) live|\bhow old\b|\bage\b|"
                       r"\brelationship|\bdating\b|\bboyfriend|\bgirlfriend|\bmoney\b|\bsalary\b|\bincome\b|"
                       r"\bgrades?\b|\bgpa\b", re.IGNORECASE)
TOPIC_WORDS = {  # what counts as a question "about" each shared interest
    "Gaming": ["gam", "video game", "console", "esport", "controller", "multiplayer", "rpg", "nintendo", "xbox", "play",
               "playstation", "steam", "level"],
    "Music": ["music", "song", "band", "concert", "album", "playlist", "artist", "singer", "listen", "instrument",
              "guitar", "piano", "genre"],
    "Photography": ["photo", "camera", "lens", "picture", "shot", "shoot", "snap", "captur", "portrait",
                    "astrophoto"],
    "Outdoors": ["outdoor", "hike", "hiking", "trail", "nature", "camp", "park", "prairie", "outside", "walk",
                 "landscape", "sunrise", "sunset"],
    "Basketball": ["basketball", "hoop", "nba", "court", "dunk", "team", "player", "shoot", "play", "sport",
                   "game"],
    "Reading": ["read", "book", "novel", "author", "chapter", "story", "library", "fiction"],
    "Movies": ["movie", "film", "cinema", "director", "actor", "theater", "theatre", "watch"],
    "Art": ["art", "paint", "draw", "sketch", "museum", "gallery", "creative"],
    "Food": ["food", "eat", "dish", "cuisine", "restaurant", "meal", "cook", "snack", "taste", "dessert", "recipe",
             "coffee", "flavor"],
    "Travel": ["travel", "trip", "visit", "country", "city", "destination", "abroad", "vacation", "explore"],
}


def first_json_object(text):
    """The first complete {...} in a reply, or None (as in A6 and connect/ai_search.py)."""
    start, tries = text.find("{"), 0
    while start != -1 and tries < 20:
        tries += 1
        depth, in_string, escaped = 0, False, False
        for end in range(start, len(text)):
            char = text[end]
            if in_string:
                if escaped:
                    escaped = False
                elif char == "\\":
                    escaped = True
                elif char == '"':
                    in_string = False
            elif char == '"':
                in_string = True
            elif char == "{":
                depth += 1
            elif char == "}":
                depth -= 1
                if depth == 0:
                    try:
                        value = json.loads(text[start:end + 1])
                    except (ValueError, RecursionError):
                        break
                    return value if isinstance(value, dict) else None
        start = text.find("{", start + 1)
    return None


def whole_json(text):
    """The reply as a JSON object if it is nothing but one, else None."""
    try:
        value = json.loads(text.strip())
    except ValueError:
        return None
    return value if isinstance(value, dict) else None


PLAIN = str.maketrans({"‐": "-", "‑": "-", "‒": "-", "–": "-", "—": "-", "−": "-",
                       "‘": "'", "’": "'", "“": '"', "”": '"', " ": " "})


def plain(text):
    """Typographic dashes, quotes and spaces made plain, so "no‑show" (a non-breaking hyphen) matches "no-show"."""
    return unicodedata.normalize("NFKC", text).translate(PLAIN)


def words(text):
    return re.findall(r"[A-Za-z0-9][A-Za-z0-9'-]*", plain(text))


def sentences(text):
    return [s for s in re.split(r"(?<=[.!?])\s+", text.strip()) if words(s)]


def has_any(text, options):
    low = plain(text).lower()
    return any(option in low for option in options)


def mentions(text, interest):
    low = plain(text).lower()
    return any(re.search(rf"\b{re.escape(word)}", low) for word in TOPIC_WORDS[interest] + [interest.lower()])


def share(checks):
    return sum(checks.values()) / len(checks) if checks else 0.0


# --- accuracy against the reference ---
def as_names(value):
    if value in (None, "", []):
        return []
    return [str(v) for v in value] if isinstance(value, list) else [str(value)]


def filter_pairs(obj):
    """Every filter a reply sets, as (key, value) pairs: one per interest, and any key counts, even an unknown one."""
    found = set()
    for key, value in obj.items():
        if key == "interests":
            found |= {("interests", name) for name in as_names(value)}
        elif value not in (None, "", []):
            found.add((key, json.dumps(value, sort_keys=True)))
    return found


def f1(predicted, expected):
    if not predicted and not expected:
        return 1.0
    hits = len(predicted & expected)
    return 0.0 if hits == 0 else 2 * hits / (len(predicted) + len(expected))


def value_errors(obj):
    errors = []
    for key, value in obj.items():
        if value is None:
            continue
        if key == "interests":
            if not isinstance(value, list) or any(v not in INTERESTS for v in value):
                errors.append(key)
        elif key == "college":
            if value not in COLLEGES:
                errors.append(key)
        elif key in CHOICES and value not in CHOICES[key]:
            errors.append(key)
    return errors


def score_extraction(reply, ref):
    obj, whole = first_json_object(reply), whole_json(reply)
    found = obj or {}
    adherence = {"only the six allowed keys": obj is not None and all(k in FIELDS for k in found),
                 "only allowed values": obj is not None and not value_errors(found),
                 "at most 4 interests": obj is not None and len(as_names(found.get("interests"))) <= 4,
                 "nothing but the JSON object": whole is not None}
    structure = 1.0 if whole is not None else 0.5 if obj is not None else 0.0
    accuracy = f1(filter_pairs(found), filter_pairs(ref)) if obj is not None else 0.0
    return accuracy, adherence, structure


def score_classification(reply, ref):
    obj, whole = first_json_object(reply), whole_json(reply)
    found = obj or {}
    decision, reason = str(found.get("decision", "")).strip(), str(found.get("reason", "")).strip()
    adherence = {"decision is APPROVE, REVIEW or REJECT": decision in LABELS,
                 "a reason of 1 to 15 words": 0 < len(words(reason)) <= 15,
                 "nothing but the JSON object": whole is not None}
    structure = (1.0 if whole is not None and set(whole) == {"decision", "reason"}
                 else 0.5 if obj is not None else 0.0)
    accuracy = 1.0 if decision.upper() == ref["decision"] else 0.0
    return accuracy, adherence, structure


def score_prediction(reply, ref):
    obj, whole = first_json_object(reply), whole_json(reply)
    found = obj or {}
    rating, stay = found.get("rating"), found.get("stay_connected")
    adherence = {"rating is a whole number from 1 to 5": isinstance(rating, int) and not isinstance(rating, bool)
                 and 1 <= rating <= 5,
                 "stay_connected is true or false": isinstance(stay, bool),
                 "nothing but the JSON object": whole is not None}
    structure = (1.0 if whole is not None and set(whole) == {"rating", "stay_connected"}
                 else 0.5 if obj is not None else 0.0)
    try:
        gap = abs(int(rating) - ref["rating"])
    except (TypeError, ValueError):
        gap = None
    rating_score = 1.0 if gap == 0 else 0.5 if gap == 1 else 0.0
    stay_value = stay if isinstance(stay, bool) else {"true": True, "false": False}.get(str(stay).strip().lower())
    stay_score = 1.0 if stay_value is not None and stay_value == ref["stay_connected"] else 0.0
    accuracy = 0.6 * rating_score + 0.4 * stay_score
    return accuracy, adherence, structure


def score_summarization(reply, ref):
    text = reply.strip()
    is_list = bool(re.search(r"^\s*([-*•]|\d+[.)])\s", text, re.MULTILINE))
    is_json = text.startswith("{")
    count = len(sentences(text))
    adherence = {"one sentence": count == 1,
                 "at most 30 words": 0 < len(words(text)) <= 30,
                 "no people's names": not any(re.search(rf"\b{name}\b", text) for name in ref["names"]),
                 "plain text: no list, quotes, markdown or preamble": bool(text) and not is_list and not is_json
                 and not re.search(r'["“”*#`]', text) and not CHATTER.search(text)}
    structure = 0.0 if not text or is_list or is_json else 1.0 if count == 1 else 0.5
    facts = ref["facts"].values()
    accuracy = sum(has_any(text, options) for options in facts) / len(facts)
    return accuracy, adherence, structure


def questions_in(reply):
    lines = [line.strip() for line in reply.strip().splitlines() if line.strip()]
    numbered = [re.match(r"^\**(\d)[.)]\**\s*(.+)$", line) for line in lines]
    found = [m.group(2).strip() for m in numbered if m]
    return lines, numbered, found or [line for line in lines if line.endswith("?")]


def score_generation(reply, ref):
    lines, numbered, questions = questions_in(reply)
    in_order = [m.group(1) for m in numbered if m] == ["1", "2", "3"]
    adherence = {"exactly 3 questions": len(questions) == 3,
                 "each ends with a question mark": bool(questions) and all(q.rstrip("*").endswith("?")
                                                                           for q in questions),
                 "each has at most 20 words": bool(questions) and all(len(words(q)) <= 20 for q in questions),
                 "no personal-detail questions": not SENSITIVE.search(reply),
                 "nothing but the 3 questions": len(lines) == 3 and len(questions) == 3}
    structure = (1.0 if len(lines) == 3 and all(numbered) and in_order
                 else 0.5 if len(questions) >= 3 else 0.0)
    asked = questions[:3]
    on_topic = sum(any(mentions(q, i) for i in ref["interests"]) for q in asked) / 3
    covered = sum(any(mentions(q, i) for q in asked) for i in ref["interests"]) / len(ref["interests"])
    accuracy = 0.5 * on_topic + 0.5 * covered
    return accuracy, adherence, structure


SCORERS = {"structured extraction": score_extraction, "classification": score_classification,
           "prediction": score_prediction, "summarization": score_summarization,
           "text generation": score_generation}


def clarity_checks(reply, kind, finished):
    text = reply.strip()
    if kind in ("summarization", "text generation"):
        ends_well = text.rstrip("*").endswith("?") if kind == "text generation" else text.endswith((".", "!", "?"))
    else:
        ends_well = first_json_object(text) is not None
    parts = [p.strip().lower() for p in re.split(r"\n|(?<=[.!?])\s+", text) if p.strip()]
    checks = {"no preamble or sign-off": bool(text) and not CHATTER.search(text),
              "not cut off": bool(text) and finished and ends_well,
              "no repeated lines": bool(parts) and len(parts) == len(set(parts)),
              "no template tokens or garbled text": bool(text) and not JUNK.search(text)}
    if kind in ("summarization", "text generation"):
        lengths = [len(words(s)) for s in sentences(text)] or [0]
        checks["readable sentence length"] = bool(text) and sum(lengths) / len(lengths) <= 30
    return checks


def evaluate(prompt, reply, finished=True):
    """All four metrics (0-1), the 1-10 score, and the rules a reply broke."""
    reply = reply or ""
    if not reply.strip():  # no answer at all: every metric is 0
        return {"accuracy": 0.0, "adherence": 0.0, "structure": 0.0, "clarity": 0.0, "score": 1.0,
                "broken rules": ["no reply"]}
    accuracy, adherence, structure = SCORERS[prompt["type"]](reply, prompt["ref"])
    clarity = clarity_checks(reply, prompt["type"], finished)
    metrics = {"accuracy": accuracy, "adherence": share(adherence), "structure": structure,
               "clarity": share(clarity)}
    form = (metrics["adherence"] + metrics["structure"] + metrics["clarity"]) / 3
    score = 1 + 9 * metrics["accuracy"] * (0.4 + 0.6 * form)
    broken = [rule for rule, ok in {**adherence, **clarity}.items() if not ok]
    return {**{m: round(v, 3) for m, v in metrics.items()}, "score": round(score, 1), "broken rules": broken}


# Self-check: an ideal reply of each type scores 10.0, an empty one 1.0, and partial failures land in between.
IDEAL = {"X1": '{"interests": ["Basketball"], "connection": "SQUAD", "day": "SATURDAY", "time": "10-12"}',
         "X5": "{}",
         "C3": '{"decision": "REJECT", "reason": "A bar at night is a nightlife venue, not a daytime meetup spot."}',
         "P4": '{"rating": 4, "stay_connected": true}',
         "S2": "The meetup failed because one person never showed up, which left the others disappointed.",
         "G1": "1. What video game could you play forever?\n2. Which song is stuck in your head this week?\n"
               "3. Do you listen to music while you are gaming?"}
FLAWED = [("P4", 'Sure! ```json\n{"rating": 4, "stay_connected": true}\n```'),  # right answer, wrong wrapping
          ("S2", "Jordan never showed up, so the meetup was a disappointment."),  # right facts, but a name
          ("C3", '{"decision": "APPROVE", "reason": "Looks fine."}'),  # right format, wrong answer
          ("G1", "Here are some questions:\n- What games do you like?\n- Favorite band?")]  # several rules broken
checks = [evaluate(PROMPT_BY_ID[pid], reply)["score"] == 10.0 for pid, reply in IDEAL.items()]
checks += [evaluate(p, "")["score"] == 1.0 for p in PROMPTS]
fenced, named, wrong, messy = (evaluate(PROMPT_BY_ID[pid], reply) for pid, reply in FLAWED)
checks.append(fenced["accuracy"] == 1.0 and fenced["structure"] == 0.5 and fenced["score"] < 10)
checks.append("no people's names" in named["broken rules"] and named["accuracy"] == 1.0)
checks.append(wrong["accuracy"] == 0.0 and wrong["structure"] == 1.0 and wrong["score"] == 1.0)
checks.append(messy["structure"] == 0.0 and messy["score"] < 7)
print("scorer self-check:", "passed" if all(checks) else f"FAILED {checks}")
display(pd.DataFrame([{"id": pid, "reply": reply.replace("\n", " / ")[:80], **evaluate(PROMPT_BY_ID[pid], reply)}
                      for pid, reply in [*IDEAL.items(), *FLAWED]]))
````

## Step 1.5: The benchmark (15 models × 25 prompts = 375 replies)

**How each reply is produced.** The messages are the type's system message, its one worked example, then the prompt.
Decoding is greedy (no sampling), so a re-run gives the same text; each type has a cap on new tokens (Step 1.2), and
the JSON types stop at the first `}`. Each local model answers one warm-up prompt (the worked example) after loading,
which is not timed, so the first measured reply does not pay for warming caches.

**What is timed.**

* **Latency**: wall-clock seconds from sending the prompt to the complete reply. For the hosted models this
  includes the network round trip and Groq's queue.
* **Tokens/sec**: generation speed. Locally, the tokens after the first one divided by the time after the first
  token (so it is not mixed up with reading the prompt); **time to first token** is reported separately. A reply
  of a single token (`{}` for the off-topic request) has no "after the first token", so its tokens/sec is output
  tokens divided by latency. For the hosted models, Groq reports its own generation time, and tokens/sec is
  completion tokens divided by it. gpt-oss is a reasoning model, so its completion tokens include its hidden
  reasoning (with `reasoning_effort: low`).

Every reply is appended to `results/runs.jsonl`, keyed by model, prompt and a hash of the exact messages and
settings, so a re-run only generates what is missing and a changed prompt never reuses an old answer.

````python
RESULTS = Path("results") / "runs.jsonl"
RESULTS.parent.mkdir(exist_ok=True)
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
COMPUTE_DTYPE = torch.bfloat16 if DEVICE == "cuda" else torch.float32
GROQ_URL = "https://api.groq.com/openai/v1/chat/completions"
HOSTED_MAX_TOKENS = 1024  # room for gpt-oss's hidden reasoning as well as the answer


def messages_for(prompt):
    kind = TYPES[prompt["type"]]
    example_in, example_out = kind["example"]
    return [{"role": "system", "content": kind["system"]}, {"role": "user", "content": example_in},
            {"role": "assistant", "content": example_out}, {"role": "user", "content": prompt["text"]}]


def run_key(spec, prompt):
    kind = TYPES[prompt["type"]]
    settings = {"messages": messages_for(prompt), "max_new_tokens": kind["max_new_tokens"], "stop": kind["stop"],
                "hosted_max_tokens": HOSTED_MAX_TOKENS if spec["plan"] == "hosted" else None}
    digest = hashlib.sha256(json.dumps(settings, sort_keys=True).encode()).hexdigest()[:12]
    return f"{spec['id']}|{prompt['id']}|{digest}"


def read_cache():
    cache = {}
    if RESULTS.exists():
        for line in RESULTS.read_text(encoding="utf-8").splitlines():
            if line.strip():
                row = json.loads(line)
                cache[row["key"]] = row
    return cache


CACHE = read_cache()


def remember(row):
    CACHE[row["key"]] = row
    with RESULTS.open("a", encoding="utf-8", newline="\n") as out:
        out.write(json.dumps(row) + "\n")


class FirstTokenClock(BaseStreamer):
    """generate() hands the prompt to put() first, then each new token: the second call is the first token."""

    def __init__(self):
        self.stamps = []

    def put(self, value):
        self.stamps.append(time.perf_counter())

    def end(self):
        pass


def load_local(spec):
    started = time.perf_counter()
    tokenizer = AutoTokenizer.from_pretrained(spec["id"])
    if spec["plan"] == "nf4":
        quant = BitsAndBytesConfig(load_in_4bit=True, bnb_4bit_quant_type="nf4", bnb_4bit_compute_dtype=COMPUTE_DTYPE)
        model = AutoModelForCausalLM.from_pretrained(spec["id"], quantization_config=quant, device_map=DEVICE)
    else:
        model = AutoModelForCausalLM.from_pretrained(spec["id"], dtype=COMPUTE_DTYPE).to(DEVICE)
    model.eval()
    return tokenizer, model, {"load_s": round(time.perf_counter() - started, 1),
                              "weights_gb": round(model.get_memory_footprint() / 1e9, 2)}


def generate_local(tokenizer, model, messages, max_new_tokens, stop):
    batch = tokenizer.apply_chat_template(messages, add_generation_prompt=True, return_tensors="pt",
                                          return_dict=True).to(model.device)
    prompt_tokens = batch["input_ids"].shape[1]
    pad = tokenizer.pad_token_id if tokenizer.pad_token_id is not None else tokenizer.eos_token_id
    clock = FirstTokenClock()
    options = {"max_new_tokens": max_new_tokens, "do_sample": False, "pad_token_id": pad, "streamer": clock}
    if stop:
        options.update(stop_strings=stop, tokenizer=tokenizer)
    started = time.perf_counter()
    with torch.inference_mode():
        output = model.generate(**batch, **options)
    latency = time.perf_counter() - started
    new_tokens = output[0][prompt_tokens:]
    reply = tokenizer.decode(new_tokens, skip_special_tokens=True)
    first = clock.stamps[1] - started if len(clock.stamps) > 1 else latency
    count = len(new_tokens)
    speed = (count - 1) / (latency - first) if count > 1 and latency > first else None
    finished = count < max_new_tokens or bool(stop and reply.rstrip().endswith(tuple(stop)))
    return reply, {"latency_s": round(latency, 3), "ttft_s": round(first, 3), "tokens_per_s": speed and round(speed, 2),
                   "new_tokens": count, "prompt_tokens": prompt_tokens, "finished": finished}


def generate_hosted(spec, messages):
    body = {"model": HOSTED[spec["id"]], "messages": messages, "temperature": 0, "max_tokens": HOSTED_MAX_TOKENS}
    if "gpt-oss" in spec["id"]:
        body["reasoning_effort"] = "low"
    for _ in range(8):  # on a rate limit (HTTP 429), wait as told and try again
        started = time.perf_counter()
        response = requests.post(GROQ_URL, json=body, timeout=120, headers={"Authorization": f"Bearer {GROQ_API_KEY}"})
        latency = time.perf_counter() - started
        if response.status_code != 429:
            break
        time.sleep(float(response.headers.get("retry-after", 10)) + 1)
    response.raise_for_status()
    data = response.json()
    choice, usage = data["choices"][0], data.get("usage") or {}
    tokens, seconds = usage.get("completion_tokens"), usage.get("completion_time")
    return choice["message"].get("content") or "", {
        "latency_s": round(latency, 3), "ttft_s": None, "tokens_per_s": round(tokens / seconds, 1) if tokens and seconds
        else None, "new_tokens": tokens, "prompt_tokens": usage.get("prompt_tokens"),
        "finished": choice.get("finish_reason") == "stop",
        "reasoning_tokens": (usage.get("completion_tokens_details") or {}).get("reasoning_tokens")}


def run_model(spec):
    """All 25 prompts for one model; cached replies are reused. Returns the rows in prompt order."""
    keyed = [(prompt, run_key(spec, prompt)) for prompt in PROMPTS]
    todo = [(prompt, key) for prompt, key in keyed if key not in CACHE]
    if todo and spec["plan"] == "hosted" and not GROQ_API_KEY:
        print(f"{spec['name']:28} skipped: set GROQ_API_KEY in ../.env to run the Very Large category")
    elif todo:
        model = tokenizer = None
        load = {"load_s": None, "weights_gb": None}
        try:
            if spec["plan"] != "hosted":
                tokenizer, model, load = load_local(spec)
                warm = TYPES["prediction"]
                generate_local(tokenizer, model, [{"role": "system", "content": warm["system"]},
                                                  {"role": "user", "content": warm["example"][0]}], 8, None)
            for prompt, key in todo:
                kind = TYPES[prompt["type"]]
                if spec["plan"] == "hosted":
                    reply, stats = generate_hosted(spec, messages_for(prompt))
                else:
                    reply, stats = generate_local(tokenizer, model, messages_for(prompt), kind["max_new_tokens"],
                                                  kind["stop"])
                remember({"key": key, "model": spec["id"], "prompt_id": prompt["id"], "type": prompt["type"],
                          "reply": reply, **stats, **load, "plan": spec["plan"]})
        except Exception as error:  # out of memory, network, provider: report it and carry on
            print(f"{spec['name']:28} FAILED: {type(error).__name__}: {str(error)[:160]}")
        finally:
            del model, tokenizer
            gc.collect()
    rows = [CACHE[key] for _, key in keyed if key in CACHE]
    if rows:
        latency = sorted(r["latency_s"] for r in rows)[len(rows) // 2]
        print(f"{spec['name']:28} {len(rows):2} replies, {len(todo) if todo else 0:2} new, median latency "
              f"{latency:6.2f} s")
    return rows
````

````python
RUNS = {spec["id"]: run_model(spec) for spec in MODELS}
````

**Timing check.** A reply generated far slower than its model's usual speed was interrupted (on a laptop: sleep,
or Windows throttling a background process), so its latency would be wrong. The check below flags any local reply
whose generation speed is under 40% of its model's median, or whose time to the first token is over 3 times the
median for the same model and prompt type. During our run two replies were caught this way, both from Falcon3-3B
while the laptop slept: C4 took 43 minutes for 26 tokens, and C5 ran at 0.3 tokens/s against about 0.9 normally.
Both were deleted from `results/runs.jsonl` and generated again (37.7 s and 39.3 s, with the same text, since
decoding is greedy), so every timing below comes from an uninterrupted reply.

````python
def stalled(rows):
    """Local replies that were interrupted: far below the model's usual speed, or a far later first token."""
    speeds = sorted(r["tokens_per_s"] for r in rows if r["tokens_per_s"])
    if not speeds or rows[0]["plan"] == "hosted":
        return []
    median_speed = speeds[len(speeds) // 2]
    first = {}
    for r in rows:
        first.setdefault(r["type"], []).append(r["ttft_s"])
    median_first = {kind: sorted(values)[len(values) // 2] for kind, values in first.items()}
    return [r for r in rows if (r["tokens_per_s"] and r["tokens_per_s"] < 0.4 * median_speed)
            or r["ttft_s"] > 3 * median_first[r["type"]]]


suspect = [row for rows in RUNS.values() for row in stalled(rows)]
print(f"timing check: {len(suspect)} interrupted replies out of {sum(len(rows) for rows in RUNS.values())}")
for row in suspect:
    print(f"  {row['model']} {row['prompt_id']}: {row['latency_s']} s, {row['tokens_per_s']} tokens/s")
````

### The benchmark table (all 375 replies)

One row per model and prompt, with the columns the assignment asks for, plus the prompt's ID (its full text is in
Step 1.3) and the four metric scores. `results/benchmark_table.csv` has the same table with the full prompt and reply
text.

````python
SPEC = {spec["id"]: spec for spec in MODELS}
records = []
for model_id, rows in RUNS.items():
    for row in rows:
        prompt = PROMPT_BY_ID[row["prompt_id"]]
        result = evaluate(prompt, row["reply"], row["finished"])
        speed = row["tokens_per_s"]
        if speed is None and row["new_tokens"]:  # a one-token reply has no "after the first token": use tokens / latency
            speed = round(row["new_tokens"] / row["latency_s"], 2)
        records.append({"Model": SPEC[model_id]["name"], "Model Category": SPEC[model_id]["category"],
                        "Tokens/sec": speed, "Latency (s)": row["latency_s"],
                        "Prompt Type": prompt["type"], "Prompt ID": prompt["id"],
                        "Prompt": prompt["text"].replace("\n", " | "),
                        "Self-Evaluation Score (1-10)": result["score"],
                        "Accuracy": result["accuracy"], "Adherence": result["adherence"],
                        "Structure": result["structure"], "Clarity": result["clarity"],
                        "Time to first token (s)": row["ttft_s"], "Output tokens": row["new_tokens"],
                        "Params (B)": SPEC[model_id]["params_b"], "Runs as": RUNS_AS[SPEC[model_id]["plan"]],
                        "Broken rules": "; ".join(result["broken rules"]), "Reply": row["reply"]})
table = pd.DataFrame(records)
table.to_csv(Path("results") / "benchmark_table.csv", index=False, lineterminator="\n")
print(f"{len(table)} rows ({table['Model'].nunique()} models x {table['Prompt ID'].nunique()} prompts); "
      "saved to results/benchmark_table.csv")
REQUIRED = ["Model", "Model Category", "Tokens/sec", "Latency (s)", "Prompt Type", "Prompt ID", "Prompt",
            "Self-Evaluation Score (1-10)", "Accuracy", "Adherence", "Structure", "Clarity"]
with pd.option_context("display.max_rows", 400, "display.max_colwidth", 90):  # full prompts: Step 1.3 and the CSV
    display(table[REQUIRED])
````

### Spot check: the scorer against our own reading

A scorer written as code is only useful if it agrees with a person. We drew 20 replies at random (4 per prompt type,
with a fixed seed), read each against its prompt and reference, and scored it from 1 to 10 by hand **before** looking
at the code's score. `results/manual_check.json` holds our scores and a one-line reason for each.

**What the check changed.** On the first comparison the code differed from us by 1.07 points on average (75% of the
replies within 1.5 points, correlation 0.75). One difference was a gap in the scorer's word lists: "engaging" did
not count as a meetup that went well. We then read every summary and icebreaker the scorer had marked down (30
replies) and found seven more wordings it missed ("improved", "a lull", "spoiled", "playing solo or with
others", "watching games live", "astrophotography", "a fictional world"). We added them and re-scored all 375
replies; the table below is after that fix.

**What it cannot see.** The remaining differences come from two limits of a rule-based scorer, so the code's scores
are, if anything, a little generous:

* it checks a classification's *decision*, not whether its *reason* is true: a correct "reject" for a bar, given
  because the model called it "a private home", scores 10 from the code and 6 from us;
* in extraction every filter counts the same, while a person weighs a missing interest (the point of the search)
  more than a missing time.

````python
sample = pd.concat([group.sample(4, random_state=7) for _, group in table.groupby("Prompt Type", sort=False)])
MANUAL = Path("results") / "manual_check.json"
manual = (pd.DataFrame(json.loads(MANUAL.read_text(encoding="utf-8"))) if MANUAL.exists()
          else pd.DataFrame(columns=["Model", "Prompt ID", "our score", "our reason"]))
check = sample[["Model", "Prompt ID", "Self-Evaluation Score (1-10)"]].merge(manual, on=["Model", "Prompt ID"],
                                                                              how="left")
check["difference"] = (check["Self-Evaluation Score (1-10)"] - check["our score"]).round(1)
with pd.option_context("display.max_colwidth", 100):
    display(check[["Model", "Prompt ID", "our score", "Self-Evaluation Score (1-10)", "difference", "our reason"]])
gap = check["difference"].abs()
agreement = check["our score"].corr(check["Self-Evaluation Score (1-10)"])
print(f"{check['our score'].notna().sum()} replies scored by hand: the code differs from us by {gap.mean():.2f} points "
      f"on average, {(gap <= 1.5).mean():.0%} are within 1.5 points, correlation {agreement:.2f}")
````

## Step 2.1: Estimate token usage

Two numbers set the token load: how many tokens one AI request uses, and how many requests a user makes in a day.

* **Tokens per request are measured.** Part 1 counted the prompt and the reply of all 375 requests, with each model's
  own tokenizer. For each job we take the mean over the 15 models. The prompts are long (the rules, a worked example,
  QuadConnect's lists) and the replies are short (a JSON object, a label, one sentence, three questions), so about 90%
  of the tokens are prompt tokens. The brief asks for the tokens *generated* per request; we give that number, and we
  also count the prompt tokens, because providers bill them and a GPU must process them before it can reply.
* **Requests per user are our estimate.** QuadConnect has no real users yet, only seed data, so there is no log to
  count from. Each of the five jobs is a page or an action in the app, and we estimate how often a daily active user
  reaches it.

````python
# How often a daily active user reaches each job: our estimate, since the app has only seed data so far.
JOBS = pd.DataFrame([
    ("structured extraction", "Search in plain words (search/ask/)", 1.0, "every active user searches once"),
    ("text generation", "Icebreakers for a match (matches/<id>/icebreakers/)", 0.5, "half of them open a match"),
    ("prediction", "Rating a feedback note after a meetup", 0.2, "a meetup about once a week"),
    ("summarization", "Staff summary of feedback (feedback/summary/)", 0.2 / 3, "one summary for every 3 notes"),
    ("classification", "Checking a newly suggested venue", 0.01, "one new venue per 100 users a day"),
], columns=["Prompt type", "QuadConnect job", "Requests per user per day", "Why"]).set_index("Prompt type")

measured = runs.groupby("type")[["prompt_tokens", "new_tokens"]].mean()  # 15 models x 5 prompts per type
JOBS["Prompt tokens"] = measured["prompt_tokens"]
JOBS["Generated tokens"] = measured["new_tokens"]
JOBS["Tokens per request"] = JOBS["Prompt tokens"] + JOBS["Generated tokens"]
JOBS["Tokens per user per day"] = JOBS["Tokens per request"] * JOBS["Requests per user per day"]

REQUESTS_PER_USER = JOBS["Requests per user per day"].sum()
PROMPT_PER_USER = round((JOBS["Prompt tokens"] * JOBS["Requests per user per day"]).sum())
GENERATED_PER_USER = round((JOBS["Generated tokens"] * JOBS["Requests per user per day"]).sum())
TOKENS_PER_USER = PROMPT_PER_USER + GENERATED_PER_USER
PROMPT_PER_REQUEST = PROMPT_PER_USER / REQUESTS_PER_USER
GENERATED_PER_REQUEST = GENERATED_PER_USER / REQUESTS_PER_USER
TOKENS_PER_REQUEST = TOKENS_PER_USER / REQUESTS_PER_USER

display(JOBS.style.format({"Requests per user per day": "{:.3f}", "Prompt tokens": "{:.1f}",
                           "Generated tokens": "{:.1f}", "Tokens per request": "{:.1f}",
                           "Tokens per user per day": "{:.1f}"}, escape="html"))  # "<id>" is text, not a tag
print(f"Average tokens per request = {TOKENS_PER_REQUEST:.0f} ({PROMPT_PER_REQUEST:.0f} prompt + "
      f"{GENERATED_PER_REQUEST:.0f} generated)")
print(f"Requests per user per day  = {REQUESTS_PER_USER:.2f}")
print(f"Average tokens per user    = {TOKENS_PER_USER} a day ({PROMPT_PER_USER} prompt + {GENERATED_PER_USER} generated)")
````

## Step 2.2: Compute total token load

The brief's formula, with our requests per user made explicit:

**Total Daily Token Load = Daily Active Users × Avg Tokens per Request × Requests per User per Day
= Daily Active Users × Avg Tokens per User**

Traffic is not even through the day: students use QuadConnect most in the afternoon and evening. Every machine count
below is sized for a busiest hour with **twice** the average hour's traffic (`PEAK = 2`), so the peak rate is the daily
load ÷ 86,400 seconds × 2.

````python
TRAFFIC = {"Prototype": 1_000, "Early Startup": 10_000, "Growing Product": 100_000, "Large Platform": 1_000_000,
           "Mass Consumer App": 10_000_000, "Global Platform": 100_000_000}
PEAK = 2  # the busiest hour has twice the average hour's traffic; machines are sized for it

load = pd.DataFrame({"Daily Active Users": TRAFFIC}).rename_axis("Daily Traffic Category")
load["Requests per day"] = load["Daily Active Users"] * REQUESTS_PER_USER
load["Avg Tokens per User"] = TOKENS_PER_USER
load["Total Daily Token Load"] = load["Daily Active Users"] * TOKENS_PER_USER
load["Generated tokens per day"] = load["Daily Active Users"] * GENERATED_PER_USER
load["Peak requests per second"] = load["Requests per day"] / DAY_S * PEAK
load["Peak tokens per second"] = load["Total Daily Token Load"] / DAY_S * PEAK
display(load.style.format("{:,.0f}").format("{:,.2f}", subset=["Peak requests per second"])
        .format("{:,.1f}", subset=["Peak tokens per second"]))
````

## Step 2.3: Hardware cost estimation (local hosting)

**The machine.** AWS's `g6.4xlarge` is the brief's machine exactly: 16 vCPUs, 64 GiB of RAM and one GPU, an NVIDIA L4
with 24 GB of memory, for $1.3232 an hour ($31.76 a day). Two models are too big for one L4 even compressed
(Qwen3.8-27B in 8-bit, gpt-oss-120b in 4-bit), so they get the `g6.12xlarge`: four L4s (96 GB) for $4.6016 an hour
($110.44 a day).

**How each model is served:** with vLLM, the usual open-source server, and never at a lower precision than Part 1
measured:

* **Under 2B: 16-bit** (bfloat16), each model's own format. Part 1 ran them in float32, which almost always gives
  the same answers.
* **2B to 15B: 8-bit (FP8).** Part 1 measured these at 4-bit, so FP8 keeps at least that quality, and the L4 does FP8
  arithmetic twice as fast as 16-bit. vLLM keeps the embeddings in 16-bit.
* **Over 15B: the format each was released in.** For gpt-oss-20b and gpt-oss-120b that is 4-bit MXFP4. For Qwen3.8-27B
  it is Qwen's own FP8 release.

**What one machine serves.** There is no L4 to measure on: our laptop has no GPU, and Part 1's three hosted models ran
on Groq's hardware. So we estimate it the standard "roofline" way, from the L4's published numbers and each model's
`config.json`. A server works in steps. In each step it reads all the weights once, writes the next token of every
request in flight, and takes in a share of the new prompts. A step lasts as long as the slower of two limits:

* **memory:** reading the weights, and every request's cache, at 300 GB/s;
* **arithmetic:** 2 operations per parameter per token, at 121 trillion a second (242 trillion in FP8).

The number of requests in flight is the memory left after the weights, divided by one request's cache (2 × layers ×
cache heads × head size × 2 bytes per token). Real servers reach only part of the datasheet peaks, so we assume 60% of
the bandwidth, 40% of the arithmetic, 3 ms of overhead per step, at most 256 requests in flight, and 80% efficiency
when one model is split over four GPUs. These are typical values for vLLM, but they are estimates: Step 2.5 shows what
changes if a machine serves half or twice as much.

**Machines Needed** = peak tokens per second ÷ what one machine serves, rounded up, and at least 1.
**Total Machine Cost** = Machines Needed × Per Machine Cost. The machines run all day, sized for the peak.

````python
# Cloud machines: AWS on-demand prices, US East (N. Virginia), Linux, from AWS's price list of 2026-10-29.
MACHINES = pd.DataFrame([
    ("g6.4xlarge", 1, 16, 64, 1.3232),   # the brief's machine: 16-core CPU, 64 GB of RAM, a GPU (one NVIDIA L4)
    ("g6.12xlarge", 4, 48, 192, 4.6016),  # four L4s, for the two models one L4 cannot hold
], columns=["Machine", "L4 GPUs", "vCPUs", "RAM (GiB)", "USD per hour"]).set_index("Machine")
MACHINES["USD per day"] = (MACHINES["USD per hour"] * 24).round(2)

L4_MEMORY_GB, L4_BANDWIDTH_GB_S = 24, 300   # NVIDIA L4 datasheet
L4_TFLOPS = {"bf16": 121, "fp8": 242}        # dense: the datasheet's 242 and 485 assume sparsity
BANDWIDTH_SHARE, ARITHMETIC_SHARE = 0.6, 0.4  # how much of those peaks a real server reaches
MULTI_GPU_SHARE = 0.8                        # a model split over 4 GPUs loses time to their traffic
STEP_OVERHEAD_S = 0.003                      # scheduling and sampling, per step
MAX_IN_FLIGHT = 256                          # vLLM's default limit on requests in flight
USABLE_GB_PER_GPU = 0.9 * L4_MEMORY_GB - 1.5  # vLLM takes 90% of the memory; 1.5 GB of that goes to activations

# From each model's config.json: layers that cache every token, layers that cache only a 128-token window, cache
# heads and head size, the embeddings (vocabulary x hidden size, once if tied), and parameters used per token.
ARCH = pd.DataFrame([
    ("SmolLM2-135M-Instruct", 30, 0, 3, 64, 49152, 576, True, 0.135),
    ("SmolLM2-360M-Instruct", 32, 0, 5, 64, 49152, 960, True, 0.362),
    ("Qwen2.5-0.5B-Instruct", 24, 0, 2, 64, 151936, 896, True, 0.494),
    ("Qwen2.5-1.5B-Instruct", 28, 0, 2, 128, 151936, 1536, True, 1.544),
    ("Falcon3-1B-Instruct", 18, 0, 4, 256, 131072, 2048, False, 1.669),
    ("granite-3.1-2b-instruct", 40, 0, 8, 64, 49155, 2048, True, 2.534),
    ("Qwen2.5-3B-Instruct", 36, 0, 2, 128, 151936, 2048, True, 3.086),
    ("Falcon3-3B-Instruct", 22, 0, 4, 256, 131072, 3072, False, 3.228),
    ("Phi-4-mini-instruct", 32, 0, 8, 128, 200064, 3072, True, 3.836),
    ("Qwen2.5-7B-Instruct", 28, 0, 4, 128, 152064, 3584, False, 7.616),
    ("Llama-3.1-Tulu-3-8B", 32, 0, 8, 128, 128264, 4096, False, 8.030),
    ("Falcon3-10B-Instruct", 40, 0, 4, 256, 131072, 3072, False, 10.306),
    ("gpt-oss-20b", 12, 12, 8, 64, 201088, 2880, False, 3.61),   # mixture of experts: 3.6B of its 20.9B per token
    ("Qwen3.8-27B", 16, 0, 4, 256, 248320, 5120, False, 27.8),  # plus 48 linear-attention layers (fixed state below)
    ("gpt-oss-120b", 18, 18, 8, 64, 201088, 2880, False, 5.13),  # mixture of experts: 5.1B of its 116.8B per token
], columns=["Model", "Full-cache layers", "Window layers", "Cache heads", "Head size", "Vocabulary", "Hidden size",
            "Tied embeddings", "Active params (B)"]).set_index("Model")
# Qwen3.8-27B's linear-attention layers keep a fixed state per request, not a cache per token:
# 48 layers x 48 heads x 128 x 128 float32 numbers, plus a small convolution state (48 x 10,240 x 3 16-bit numbers).
ARCH["Fixed state per request (MB)"] = 0.0
ARCH.loc["Qwen3.8-27B", "Fixed state per request (MB)"] = (48 * 48 * 128 * 128 * 4 + 48 * 10_240 * 3 * 2) / 1e6
RELEASED = {"gpt-oss-20b": ("MXFP4", 13.76), "Qwen3.8-27B": ("FP8", 30.87),  # GB of the released weight files
            "gpt-oss-120b": ("MXFP4", 65.25)}


def serving(name):
    """How one model is served: precision, weights in GB, the arithmetic it runs, and its machine (rules above)."""
    params, arch = MODEL_INFO.loc[name, "Params (B)"], ARCH.loc[name]
    embeddings = arch["Vocabulary"] * arch["Hidden size"] * (1 if arch["Tied embeddings"] else 2) / 1e9
    if name in RELEASED:
        precision, weights_gb = RELEASED[name]
    elif params < 2:
        precision, weights_gb = "BF16", params * 2
    else:
        precision, weights_gb = "FP8", embeddings * 2 + (params - embeddings)
    machine = "g6.4xlarge" if weights_gb <= USABLE_GB_PER_GPU - 4 else "g6.12xlarge"  # one L4 if 4 GB stay for caches
    return {"Serve as": precision, "Arithmetic": "fp8" if precision == "FP8" else "bf16",  # MXFP4 unpacks to 16-bit
            "Weights (GB)": round(weights_gb, 2), "Machine": machine}


SERVE = pd.DataFrame({name: serving(name) for name in NAMES}).T


def capacity(name, prompt=PROMPT_PER_REQUEST, generated=GENERATED_PER_REQUEST, factor=1.0):
    """What one machine serves for this model, for requests of this shape (prompt and generated tokens)."""
    arch, serve = ARCH.loc[name], SERVE.loc[name]
    gpus = MACHINES.loc[serve["Machine"], "L4 GPUs"]
    share = MULTI_GPU_SHARE if gpus > 1 else 1.0
    per_token = 2 * arch["Cache heads"] * arch["Head size"] * 2  # keys and values, 2 bytes each, per layer

    def cache_bytes(tokens):  # one request's cache while it holds this many tokens
        return (per_token * (arch["Full-cache layers"] * tokens + arch["Window layers"] * min(tokens, 128))
                + arch["Fixed state per request (MB)"] * 1e6)

    free = (gpus * USABLE_GB_PER_GPU - serve["Weights (GB)"]) * 1e9
    in_flight = min(MAX_IN_FLIGHT, int(free // cache_bytes(prompt + generated)))
    tokens_per_step = in_flight * (1 + prompt / generated)  # one new token per request, plus its share of prompts
    memory_s = ((serve["Weights (GB)"] * 1e9 + in_flight * cache_bytes(prompt + generated / 2))
                / (gpus * L4_BANDWIDTH_GB_S * 1e9 * BANDWIDTH_SHARE * share))
    arithmetic_s = (2 * arch["Active params (B)"] * 1e9 * tokens_per_step
                    / (gpus * L4_TFLOPS[serve["Arithmetic"]] * 1e12 * ARITHMETIC_SHARE * share))
    step_s = max(memory_s, arithmetic_s) + STEP_OVERHEAD_S
    requests_per_s = in_flight / (generated * step_s) * factor  # factor: Step 2.5's what-ifs
    return {"Cache per request (MB)": cache_bytes(prompt + generated) / 1e6, "Requests in flight": in_flight,
            "Limit": "memory" if memory_s > arithmetic_s else "arithmetic", "Step (ms)": step_s * 1e3,
            "Requests/s": requests_per_s, "Tokens/s": requests_per_s * (prompt + generated)}


SERVE = SERVE.join(pd.DataFrame({name: capacity(name) for name in NAMES}).T)
SERVE["Per machine USD/day"] = SERVE["Machine"].map(MACHINES["USD per day"])
SERVE["Tokens per machine per day"] = SERVE["Tokens/s"] * DAY_S / PEAK  # sized for the peak, so half used on average
SERVE["USD per 1M tokens"] = SERVE["Per machine USD/day"] / SERVE["Tokens per machine per day"] * 1e6
display(MACHINES)
display(MODEL_INFO[["Model Category", "Params (B)"]].join(SERVE).style.format({
    "Params (B)": "{:.3f}", "Weights (GB)": "{:.2f}", "Cache per request (MB)": "{:.1f}", "Step (ms)": "{:.1f}",
    "Requests/s": "{:,.1f}", "Tokens/s": "{:,.0f}", "Per machine USD/day": "${:,.2f}",
    "Tokens per machine per day": "{:,.0f}", "USD per 1M tokens": "${:.4f}"}))
````

**Reading the table.** *Limit* says which wall a model hits first. All 15 hit **arithmetic**, not memory, because
our requests are mostly prompt: a server spends far more work reading prompts than writing replies, and reading a
prompt is arithmetic. *USD per 1M tokens* is what a machine's tokens cost when it is as busy as we plan (full in the
busiest hour). It is the number to hold against an API's price per token in Step 2.4.

The 90-row table: every model at every traffic level.

````python
def local_cost(name, users, tokens_per_user=TOKENS_PER_USER, tokens_per_s=None):
    """Machines needed and their daily cost to host one model for this many daily users."""
    tokens_per_s = SERVE.loc[name, "Tokens/s"] if tokens_per_s is None else tokens_per_s
    machines = max(1, math.ceil(users * tokens_per_user / DAY_S * PEAK / tokens_per_s))
    return machines, machines * SERVE.loc[name, "Per machine USD/day"]


local_rows = []
for name in NAMES:
    for level, users in TRAFFIC.items():
        machines, cost = local_cost(name, users)
        local_rows.append({"Model Name": name, "Cloud machine": SERVE.loc[name, "Machine"],
                           "Daily Traffic Category": level, "Avg Tokens per User": TOKENS_PER_USER,
                           "Total Daily Token Load": users * TOKENS_PER_USER,
                           "Per Machine Cost (Cloud), USD/day": SERVE.loc[name, "Per machine USD/day"],
                           "Machines Needed": machines, "Total Machine Cost, USD/day": round(cost, 2)})
local_costs = pd.DataFrame(local_rows)
local_costs.to_csv(RESULTS_DIR / "local_hosting_costs.csv", index=False, lineterminator="\n")
print(f"{len(local_costs)} rows ({local_costs['Model Name'].nunique()} models x "
      f"{local_costs['Daily Traffic Category'].nunique()} traffic levels); saved to results/local_hosting_costs.csv")
display(local_costs.style.hide(axis="index").format({
    "Avg Tokens per User": "{:,}", "Total Daily Token Load": "{:,}", "Per Machine Cost (Cloud), USD/day": "${:,.2f}",
    "Machines Needed": "{:,}", "Total Machine Cost, USD/day": "${:,.2f}"}))
````

## Step 2.4: Hugging Face Inference API cost comparison

Hugging Face's [Inference Providers](https://huggingface.co/docs/inference-providers) serve open models through one
API (`router.huggingface.co`), at each provider's price per token. There are no machines to rent, and the bill grows
with use. The prices are from Hugging Face's live list on 3 November 2026, in US dollars per million tokens. We kept only
live providers with a per-token price.

**The comparable paid model.** Three of our 15 are sold as they are: gpt-oss-20b, Qwen3.8-27B and gpt-oss-120b.
Llama-3.1-Tulu-3-8B is a fine-tune of one that is sold, Llama-3.1-8B-Instruct. For the rest we take the sold model
closest in size. No provider sells a model under 3B, so all six of ours under 3B map to IBM's granite-4.2-3b, the
smallest model on sale (and a later generation of our granite-3.1-2b).

**Cost of Paid API per Token.** Providers charge more for generated tokens than for prompt tokens, and our requests
are about 90% prompt. So for each model we take the provider that is cheapest **for our mix**, and we quote its blended
price: (prompt tokens × prompt price + generated tokens × generated price) ÷ all tokens.

**Machines Required (API infrastructure).** The API runs the model, but QuadConnect still needs servers that build
each prompt, call the API and run Part 1's output checks (the allow-list, the strict schema, the no-names check). We
use AWS's `m7i.large`: 2 vCPUs, 8 GiB and no GPU, for $0.1008 an hour ($2.42 a day). Every size of that family costs the
same per vCPU ($0.0504 an hour), so the smallest lets the count grow with traffic without changing the cost at scale; a
16-core server would sit almost idle at 1,000 users. We assume one handles 25 calls a second at peak (12.5 per core):
each call mostly waits on the provider. **Total Cost** = the API bill + these servers.

````python
# Hugging Face Inference Providers: (provider, USD per 1M prompt tokens, USD per 1M generated tokens), every live
# provider with a per-token price, from https://router.huggingface.co/v1/models on 2026-11-03.
HF_PRICES = {
    "ibm-granite/granite-4.2-3b": [("deepinfra", 0.03, 0.12)],
    "google/gemma-3-4b-it": [("deepinfra", 0.05, 0.10)],
    "meta-llama/Llama-3.1-8B-Instruct": [("novita", 0.02, 0.05), ("deepinfra", 0.02, 0.05)],
    "Qwen/Qwen3.5-9B": [("ovhcloud", 0.12, 0.18), ("deepinfra", 0.10, 0.15)],
    "openai/gpt-oss-20b": [("groq", 0.10, 0.50), ("ovhcloud", 0.05, 0.18), ("deepinfra", 0.03, 0.14)],
    "Qwen/Qwen3.8-27B": [("novita", 0.42, 3.00), ("cerebras", 0.99, 1.49), ("nscale", 0.20, 2.50),
                         ("ovhcloud", 0.47, 3.19), ("deepinfra", 0.20, 2.50)],
    "openai/gpt-oss-120b": [("groq", 0.15, 0.75), ("novita", 0.05, 0.25), ("cerebras", 0.35, 0.75),
                            ("baseten", 0.10, 0.50), ("ovhcloud", 0.09, 0.47), ("deepinfra", 0.037, 0.17)],
}
SMALLEST = "nothing under 3B is sold: the smallest sold model"
COMPARABLE = {  # our model: (the paid model on Hugging Face, why that one)
    "SmolLM2-135M-Instruct": ("ibm-granite/granite-4.2-3b", SMALLEST),
    "SmolLM2-360M-Instruct": ("ibm-granite/granite-4.2-3b", SMALLEST),
    "Qwen2.5-0.5B-Instruct": ("ibm-granite/granite-4.2-3b", SMALLEST),
    "Qwen2.5-1.5B-Instruct": ("ibm-granite/granite-4.2-3b", SMALLEST),
    "Falcon3-1B-Instruct": ("ibm-granite/granite-4.2-3b", SMALLEST),
    "granite-3.1-2b-instruct": ("ibm-granite/granite-4.2-3b", "the same family, a later generation"),
    "Qwen2.5-3B-Instruct": ("ibm-granite/granite-4.2-3b", "the closest size (3B)"),
    "Falcon3-3B-Instruct": ("ibm-granite/granite-4.2-3b", "the closest size (3B)"),
    "Phi-4-mini-instruct": ("google/gemma-3-4b-it", "the closest size (4B)"),
    "Qwen2.5-7B-Instruct": ("meta-llama/Llama-3.1-8B-Instruct", "the closest size (8B)"),
    "Llama-3.1-Tulu-3-8B": ("meta-llama/Llama-3.1-8B-Instruct", "its base model"),
    "Falcon3-10B-Instruct": ("Qwen/Qwen3.5-9B", "the closest size (9B)"),
    "gpt-oss-20b": ("openai/gpt-oss-20b", "the same model"),
    "Qwen3.8-27B": ("Qwen/Qwen3.8-27B", "the same model"),
    "gpt-oss-120b": ("openai/gpt-oss-120b", "the same model"),
}


def cheapest_for_our_mix(paid, prompt=PROMPT_PER_USER, generated=GENERATED_PER_USER):
    """The provider with the lowest blended price for this mix of tokens: (USD per 1M tokens, provider, in, out)."""
    return min(((prompt * p_in + generated * p_out) / (prompt + generated), provider, p_in, p_out)
               for provider, p_in, p_out in HF_PRICES[paid])  # a tie goes to the first name in alphabetical order


api_models = {}
for name in NAMES:
    paid, why = COMPARABLE[name]
    blended, provider, p_in, p_out = cheapest_for_our_mix(paid)
    api_models[name] = {"Comparable paid model": paid, "Why this one": why, "Cheapest provider for our mix": provider,
                        "USD per 1M prompt tokens": p_in, "USD per 1M generated tokens": p_out,
                        "USD per 1M tokens (our mix)": blended}
API_MODELS = pd.DataFrame(api_models).T
API_SERVER = {"Machine": "m7i.large", "USD per day": round(0.1008 * 24, 2), "Peak calls per second": 25}
display(MODEL_INFO[["Model Category", "Params (B)"]].join(API_MODELS).style.format({
    "Params (B)": "{:.3f}", "USD per 1M prompt tokens": "${:.3f}", "USD per 1M generated tokens": "${:.2f}",
    "USD per 1M tokens (our mix)": "${:.4f}"}))
````

````python
def api_cost(name, users, prompt_per_user=PROMPT_PER_USER, generated_per_user=GENERATED_PER_USER):
    """The daily API bill, the servers that call the API, and the total, for one model and this many daily users."""
    blended = cheapest_for_our_mix(COMPARABLE[name][0], prompt_per_user, generated_per_user)[0]
    bill = round(users * (prompt_per_user + generated_per_user) * blended / 1e6, 2)
    servers = max(1, math.ceil(users * REQUESTS_PER_USER / DAY_S * PEAK / API_SERVER["Peak calls per second"]))
    return bill, servers, round(bill + servers * API_SERVER["USD per day"], 2)


api_rows = []
for name in NAMES:
    info, paid = MODEL_INFO.loc[name], API_MODELS.loc[name]
    for level, users in TRAFFIC.items():
        bill, servers, total = api_cost(name, users)
        api_rows.append({"Local LLM Model Size": f"{name} ({info['Params (B)']}B, {info['Model Category']})",
                         "Comparable Paid Model Hosted on Hugging Face":
                             f"{paid['Comparable paid model']} ({paid['Cheapest provider for our mix']})",
                         "Cost of Paid API per Token, USD per 1M tokens": paid["USD per 1M tokens (our mix)"],
                         "Daily Traffic Category": level, "Total Token Load": users * TOKENS_PER_USER,
                         "Estimated API Cost, USD/day": bill, "Machines Required (API infrastructure)": servers,
                         "Total Cost, USD/day": total})
api_costs = pd.DataFrame(api_rows)
api_costs.to_csv(RESULTS_DIR / "api_costs.csv", index=False, lineterminator="\n")
print(f"{len(api_costs)} rows ({api_costs['Local LLM Model Size'].nunique()} models x "
      f"{api_costs['Daily Traffic Category'].nunique()} traffic levels); saved to results/api_costs.csv")
display(api_costs.style.hide(axis="index").format({
    "Cost of Paid API per Token, USD per 1M tokens": "${:.4f}", "Total Token Load": "{:,}",
    "Estimated API Cost, USD/day": "${:,.2f}", "Machines Required (API infrastructure)": "{:,}",
    "Total Cost, USD/day": "${:,.2f}"}))
````

## Step 2.5: Analysis

The evidence first: from how many daily users hosting becomes cheaper for each model, and how much that depends on
our estimates. The answers follow.

The break-even is searched over every number of users from 1,000 to 100 million (50 points per tenfold), and we report
the first point from which hosting stays cheaper. Four what-ifs test the estimates:

* **as estimated**: the numbers above;
* **machines serve 2× more**: if a real benchmark beats our roofline, or if machines stay busy around the clock
  instead of being sized for the peak (either halves the cost per token);
* **machines serve 2× less**: if a real benchmark falls short of our roofline;
* **each model's own tokens**: each model's own prompt and reply lengths from Part 1 instead of the average (its
  tokenizer, and its habit of writing short or long replies).

````python
def model_tokens(name):
    """This model's own prompt and generated tokens per user per day (Part 1's counts, Step 2.1's job mix)."""
    own = runs[runs["Model"] == name].groupby("type")[["prompt_tokens", "new_tokens"]].mean()
    mix = JOBS["Requests per user per day"]
    return round((own["prompt_tokens"] * mix).sum()), round((own["new_tokens"] * mix).sum())


def break_even(name, factor=1.0, own_tokens=False):
    """The fewest daily users from which hosting stays cheaper than the API, or None if never up to 100 million."""
    prompt, generated = model_tokens(name) if own_tokens else (PROMPT_PER_USER, GENERATED_PER_USER)
    tokens_per_s = capacity(name, prompt / REQUESTS_PER_USER, generated / REQUESTS_PER_USER, factor)["Tokens/s"]
    dearer = [i for i, u in enumerate(users_grid)
              if local_cost(name, u, prompt + generated, tokens_per_s)[1] >= api_cost(name, u, prompt, generated)[2]]
    if dearer and dearer[-1] == len(users_grid) - 1:
        return None  # hosting still costs more at 100 million users
    return users_grid[dearer[-1] + 1 if dearer else 0]


def about(users):
    """A break-even, rounded: the search steps are 5% apart, so more digits would claim too much."""
    if pd.isna(users):
        return "never"
    return f"{users / 1e6:.1f} million" if users >= 1e6 else f"{round(users, -4):,.0f}"


SCENARIOS = {"as estimated": {}, "machines serve 2x more": {"factor": 2.0}, "machines serve 2x less": {"factor": 0.5},
             "each model's own tokens": {"own_tokens": True}}
crossover = pd.DataFrame({label: {name: break_even(name, **args) for name in NAMES}
                          for label, args in SCENARIOS.items()})
summary = MODEL_INFO[["Model Category", "Part 1 score"]].join(SERVE[["USD per 1M tokens"]]).join(
    API_MODELS[["USD per 1M tokens (our mix)"]].rename(columns={"USD per 1M tokens (our mix)": "API USD per 1M tokens"}))
summary = summary.rename(columns={"USD per 1M tokens": "Local USD per 1M tokens (busy machine)"})
summary["Local ÷ API, per token"] = summary["Local USD per 1M tokens (busy machine)"] / summary["API USD per 1M tokens"]
shown = crossover.map(about)
display(summary.join(shown).style.format({"Part 1 score": "{:.2f}", "Local USD per 1M tokens (busy machine)": "${:.4f}",
                                          "API USD per 1M tokens": "${:.4f}", "Local ÷ API, per token": "{:.2f}"}))
for label in SCENARIOS:
    print(f"{label:26} hosting wins for {crossover[label].notna().sum():2} of 15 models by 100 million users")
````

**1. At what scale is local hosting cheaper?**

Only from about **1 million daily users**, and only for the smallest models. As estimated, hosting is cheaper in
14 of the 90 cases. Five models end up cheaper to host: the three Ultra-Light models from about 1 million daily users,
granite-3.1-2b from about 6.3 million, and Phi-4-mini from about 25 million. For the other ten it is still dearer at
100 million. Machines come in whole units, so near a break-even the cheaper option can switch back and forth:
granite-3.1-2b ties with the API at 1 million users, then the API is cheaper again until about 6.3 million.

The reason is the cost per token on a busy machine, in the tables above. Hosting wins only where our machine makes a
token for less than the API sells one. That happens for the Ultra-Light models, which are far smaller than the
smallest model on sale (granite-4.2-3b, $0.0388 per million tokens for our mix): their busy-machine cost is $0.0049 to
$0.0158. It almost happens for the models that match what is on sale: granite-3.1-2b is 1.01× the API's price per
token, Phi-4-mini 1.08×, Qwen3.8-27B 1.09×. It never happens for the 7B, 8B and gpt-oss models. Providers sell those
tokens 2.7 to 5.4 times cheaper than an L4 makes them: Llama-3.1-8B at $0.0229 per million, against $0.1230 for
Llama-3.1-Tulu-3-8B on our machine. Providers run on bigger GPUs that stay busy all day.

The cheap local cases come with a catch. The models that win most clearly are the Ultra-Light ones, which scored
4.51 to 6.81 in Part 1: too weak for most of QuadConnect's jobs, while the API's granite-4.2-3b is a much better model.
Among the models good enough to use (8 or more in Part 1), where hosting wins the API never costs more than 1.12
times as much.

**2. At what scale is API usage cheaper?**

**Everywhere up to 100,000 daily users, for all 15 models**, by 3.3 to 45 times. A GPU machine costs $31.76 a day
(the 4-GPU machine $110.44) even when it is nearly idle. Through the API, 1,000 users' tokens for a small model cost
$0.03 a day, plus $2.42 for the server that calls the API; 100,000 users' tokens cost $2.70. Below about 380,000
daily users the API is cheaper for every model in every what-if we ran.

Past 1 million users the API is still cheaper for **10 of the 15 models at every level**: Qwen2.5-1.5B,
Falcon3-1B, Qwen2.5-3B, Falcon3-3B, Qwen2.5-7B, Llama-3.1-Tulu-3-8B, Falcon3-10B and all three Very Large models.
At 100 million users its lead is 1.1 to 4.3 times, $93,069 a month against $100,044 for Qwen2.5-3B, for example.

**How sure can we be?** The prices are real. The GPU capacity is our estimate, so we tested it:

* If our machines serve **2× more** than estimated, hosting wins for 11 of the 15 models, from 0.7 to 2 million
  users. A real benchmark could show this, and so could machines that are busy around the clock instead of sized for the
  peak.
* If they serve **2× less**, only the three Ultra-Light models win.
* With **each model's own token counts** instead of the average, the same five models win.
* Never, in any what-if, does hosting win for Qwen2.5-7B, Llama-3.1-Tulu-3-8B, gpt-oss-20b or gpt-oss-120b.

So the conclusion that does not move is this: below a few hundred thousand daily users, rent tokens, not machines.

**3. When would you switch your architecture?**

Part 1 gave us three tiers: a fast local model (Qwen2.5-1.5B), a quality local model (Qwen2.5-3B) and an API model
(Qwen3.8-27B). Part 2 says when each should really run on our own machines:

1. **Start on the API, for every job.** Use Hugging Face's router with the cheapest provider for our mix. Up to
   100,000 daily users the bill is $74 to $154 a month for the small models, against $953 for one GPU machine. There is
   also nothing to operate.
2. **Reconsider at about 1 million daily users.** That is where one machine's daily cost meets the API bill for the
   small models ($31.76 against $31.87). The decision then rests on a number we could not measure. So the trigger is to
   **benchmark one `g6.4xlarge`** with vLLM on our real prompts. A machine must serve at least 1.23× our estimate
   for Qwen2.5-3B (about 19,000 tokens a second) to beat the API per token.
3. **Move only the high-volume, small-model jobs.** Search alone is 74% of our tokens, and icebreakers are 18%.
   Those can run on Qwen2.5-3B or granite-3.1-2b machines once the benchmark clears the bar. Everything else stays on the API.
4. **Keep the large models on the API at every scale.** Providers sell gpt-oss tokens 2.7 to 3.4 times cheaper than
   our L4 machines can make them. Hosting those would need bigger GPUs, which is a different machine and a different
   calculation.

Some reasons can bring the switch earlier or push it later, and they are not on the bill:

* **Earlier:** feedback notes are students' own words, and on our machines they never leave our account. Only one
  provider sells granite-4.2-3b, so an outage stops the feature. Provider rate limits can cap a busy day.
* **Later:** GPU machines need monitoring, updates and someone on call. A real deployment also needs a second machine
  for failover, which doubles the smallest local bill. Where hosting does win among the good models, it saves little:
  granite-3.1-2b saves $768 a month at 10 million users, less than the engineering time it takes.
